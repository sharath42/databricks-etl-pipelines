Creating the Catalog and Schema


```python
%sql

CREATE CATALOG IF NOT EXISTS cdc_catalog;
```

```python
%sql
Use Catalog cdc_catalog;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS cdc_catalog.bronze;
CREATE SCHEMA IF NOT EXISTS cdc_catalog.silver;
CREATE SCHEMA IF NOT EXISTS cdc_catalog.gold;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS cdc_catalog.cdc_demo;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS cdc_catalog.cdc_demo.cdc_files;
```

```python
%sql
CREATE TABLE IF NOT EXISTS cdc_catalog.silver.silver_customer
(
    customer_id BIGINT,
    name STRING,
    city STRING,
    salary DECIMAL(12,2),
    updated_at TIMESTAMP
)
USING DELTA;
```
